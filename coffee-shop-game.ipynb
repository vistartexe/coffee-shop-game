```python
# importing functions

# HTML : lets display HTML code in the output
# clear_output : clears the output of the current cell.

from IPython.display import HTML, clear_output # importing HTML and clear_output functions from IPython.display module and IPython library

from tabulate import tabulate # importing tabulate function from tabulate module; used for generating nicely formatted tables

import time # importing time module; used for getting the current time, pausing or delaying code execution, measuring the execution time of code blocks, and working with timestamps.

import random # importing random module; used for generating random numbers and making random choices


player_bucks = 500 # stores the user's bucks with the default value of 500

day = 1 # stores the user's progress as in day

day_skipped = 0 # stores the user's total of day skipped

customers_served = 0 # stores total amount of customers served

read_prologue = False # stores bool value whether the user's already read the prologue or not


# user's Action Function
def user_action(heading, additional, total, select1, select2="", select3="", select4="", select5=""):

    while True:
        
        selection = [] # stores how many selection(s) will be shown soon

        # stores the parameters
        select = [select1, select2, select3, select4, select5]

        # heading text HTML code
        if heading != False:
            display(HTML(f"<h1>{heading}</h1>"))

        # additional text
        if additional != False:
            print(f"{additional}\n")

        # adding what was in the parameters into the selection list
        for i in range(0, total):
            selection.append(f"[{i+1}] {select[i]}")
            print(selection[i])

        # ensures that user inputted an integer
        try:
            user_select = int(input("\nACTION: "))
            
            if user_select in range(1, total+1):
                print(select[user_select-1])
                time.sleep(1)
                clear_output()
                return user_select

            else:
                print("Invalid Output")
                time.sleep(1)
                clear_output()

        # if the user inputted string, ValueError is the type of error shown
        except ValueError:
            print("Invalid Output")
            time.sleep(1)
            clear_output()

# the opening screen and prologue function
def open_and_prologue():

    # declares the variable as having a global scope aka allows to modify it outside of its normal scope
    # also referring to the global variable (read_prologue), not creating a new local variable with the same name
    global read_prologue

    # opening screen
    user_action("Coffee Shop Saga: 7 Days of Summer Brew", False, 1, "PLAY")
    
    # prologue Story
    prologue = (
        """During summertime after you've done your semester exam, as a high school student, you were thrilled to
spend your holiday at your grandmother’s quaint coffee shop; a place filled with the aroma of fresh pastries and the comforting
hum of friendly chatter. But upon your arrival, the cheerful atmosphere was overshadowed by a grim announcement:
The coffee shop was planned for a demolition in just seven days to make way for a new shopping complex.""",
        """Your grandmother’s coffee shop had been a cornerstone of the community for decades,
a place where locals gathered to share stories, laughter, and the perfect cup of coffee.
The news hit you hard; this wasn’t just a business, it was a sanctuary filled with memories of childhood summers spent helping
your grandmother behind the counter. The city council delegates have proposed a compensation for the trouble
but your grandmother rejected it immediately.""",
        """Determined to save this beloved spot, you quickly rally your friends and the loyal customers who truly cherish the coffee shop.
With only a week to prove the coffee shop's worth to the city council, you brainstorm creative ideas to showcase the coffee shop's importance.""",
        """As you step into the coffee shop on the first day, you can almost hear your grandmother’s voice encouraging you.
The clock is ticking, and you feel the weight of the task ahead. Can you ignite a passion in the community and possibly create a campaign
to save the coffee shop before it’s too late? The summer may be fleeting, but the fight for the coffee shop has just begun."""
    )

    prologue_part = 1 # stores which current part is being read by user's

    if read_prologue == False:
        for story in prologue:
            prologue_head = f"Prologue ({prologue_part}/4)"
            user_action(prologue_head, story, 1, "NEXT")
            prologue_part += 1

    read_prologue = True

# default day 1 menu
MENU = {
    "Espresso": 25,
    "Americano": 20,
    "Latte": "LOCKED",
    "Cappuccino": "LOCKED",
    "Mocha": "LOCKED",
    "Croissant": 30,
    "Muffin": "LOCKED",
    "Fruit Tart": "LOCKED"
}

# event at the end of the day
def end_day_event(current_day, start_or_skip_day):

    if start_or_skip_day == True:
        if current_day == 1:
            user_action(False, "+Unlocked Latte!", 1, "NEXT")
            MENU["Latte"] = 30
    
        elif current_day == 3:
            user_action(False, "+Unlocked Muffin!", 1, "NEXT")
            MENU["Muffin"] = 35
    
        elif current_day == 5:
            user_action(False, "+Unlocked Mocha!\n+Unlocked Fruit Tart!", 1, "NEXT")
            MENU["Mocha"] = 40
            MENU["Fruit Tart"] = 35
             
        elif current_day == 6:
            user_action(False, "+Unlocked Cappuccino!", 1, "NEXT")
            MENU["Cappuccino"] = 35
            
    if current_day == 7:
        print("The long awaited day has finally come...")
        time.sleep(3)
        clear_output()
        
        if not "LOCKED" in MENU.values() and player_bucks >= 1500:

            # good ending story
            good_ending = """After a week of hard work and community support, your efforts pay off. The city council members,
moved by the outpouring of love and support, decide to visit the coffee shop themselves. Seeing the community's spirit
firsthand, they agree to reconsider the demolition plans. In a thrilling turn of events, they propose a partnership
with your grandmother to renovate the shop instead. Your grandmother beams with pride, and you realize that this summer
has not only saved a cherished place but also deepened your bonds with your community and family"""
            
            user_action("Good Ending", good_ending, 1, "NEXT")

            user_action(False, "You obtained Good Ending", 1, "NEXT")
            
        elif not "LOCKED" in MENU.values() or player_bucks >= 1500:

            # neutral ending story
            neutral_ending = """Throughout the week, you rally support, but despite your best efforts, the city council
remains unconvinced. While the community comes together for a final event, the turnout is decent but not overwhelming.
The city council proceeded with the demolition. However, they offer to help your grandmother relocate her coffee shop
to a new location, promising to preserve its essence in the new space. Your grandmother reluctantly agrees, and although
the original shop will be lost, there is a sense of hope as plans begin for a fresh start. You leave the coffee shop
knowing that while this chapter has closed, the spirit of the coffee shop will continue elsewhere,
and you’re proud of the fight you put up for something you love."""

            user_action("Neutral Ending", neutral_ending, 1, "NEXT")

            user_action(False, "You obtained Neutral Ending", 1, "NEXT")

        else:

            # bad ending story
            bad_ending = """Despite your best efforts, the community shows little enthusiasm, and the rally falls flat.
The demolition proceeds as planned, and you watch helplessly as the beloved coffee shop is torn down.
Your grandmother is heartbroken, feeling like a piece of her life has been erased. The summer that was supposed
to be filled with joy turns into a time of loss and mourning for what once was. You leave, disheartened and uncertain,
knowing that while you fought for something you loved, it wasn’t enough to change the outcome. The experience leaves
you with a bittersweet lesson about community, resilience, and the fleeting nature of cherished places."""

            user_action("Bad Ending", bad_ending, 1, "NEXT")

            user_action(False, "You obtained Bad Ending", 1, "NEXT")

        return True
        

# main menu function
def main_menu():
    exit = False

    while exit == False:

        # declaring these variables global
        global player_bucks
    
        global day
    
        global day_skipped
    
        global customers_served

        # statistic dictionary
        statistic = {
            "Day" : day,
            "Day Skipped" : day_skipped,
            "Total Bucks" : player_bucks,
            "Total Customers Served" : customers_served
        }

        # statistic function
        def statisticfunction(option):

            statistic_list = [] # stores the statistic that will be shown soon

            # loop to add the statistic
            for stat_name, stat in statistic.items():
                if stat_name == "Total Bucks":
                    statistic_list.append([stat_name, f"${stat:.2f}"])
                else:
                    statistic_list.append([stat_name, stat])

                # making the statistic list into tables
                tabulate_statistic = tabulate(statistic_list, tablefmt="fancy_grid")

            # user action in the stats menu
            user_action("Stats", tabulate_statistic, 1, option)

        # user action in main menu
        main_menu_select = user_action(f"Day {day}<br>Bucks : ${player_bucks:.2f}", False, 5, "MENU", "START DAY", "SKIP DAY", "STATS", "EXIT")

        # menu
        if main_menu_select == 1:

            menu_list_items = [] # stores the menu that will be shown soon

            # loop to add the menu
            for item, price in MENU.items():
                if price != "LOCKED":
                    menu_list_items.append([item, f"${price:.2f}"])
                else:
                    menu_list_items.append([item, "LOCKED"])

            # making the menu list into tables
            tabulate_menu = tabulate(menu_list_items, headers=["Item", "Price"], tablefmt="fancy_grid")

            # user action in the menu
            user_action("Menu", tabulate_menu, 1, "BACK")

        # start day
        elif main_menu_select == 2:
            clear_output()
            
            customers = 0 # stores amount of customer served in the day
            max_customers = random.randint(3, 10) # randomize max customers in a day
            total_earnings = 0 # stores total earnings in the day

            # display day in HTML code
            display(HTML(f"<h1>Day {day}</h1>"))

            # day start
            while customers <= max_customers:
                random_menu = random.choice(list(MENU.keys()))
    
                vowels = ["A", "I", "U", "E", "O"]
                
                if random_menu[0] not in vowels:
                    article = "a"
                else:
                    article = "an"
    
                if MENU[random_menu] != "LOCKED":
                    
                    customer_event = f"A customer bought {article} {random_menu}! +${MENU[random_menu]} Bucks"
                        
                    print(customer_event)
    
                    total_earnings += MENU[random_menu]
                    customers += 1
                    time.sleep(random.randint(0, 2))
    
            print("\nGood work! Let's wrap it up for now and call it a day!")
    
            time.sleep(3)
            clear_output()

            # day statistic dictionary
            day_statistic = {
                "Total Earnings" : total_earnings,
                "Total Customers" : customers
            }
    
            day_statistic_list = [] # stores the statistic in the day that will be shown soon

            # loop to add the statistic
            for stat_name, stat in day_statistic.items():
                if stat_name == "Total Earnings":
                    day_statistic_list.append([stat_name, f"${stat:.2f}"])
                else:
                    day_statistic_list.append([stat_name, stat])

            # making the statistic list into tables
            tabulate_day_statistic = tabulate(day_statistic_list, tablefmt="fancy_grid")

            user_action(False, tabulate_day_statistic, 1, "NEXT DAY")

            # adding total earnings in the day into user's total bucks
            player_bucks += total_earnings

            # adding total customers served in the day into user's total customers served
            customers_served += customers

            # runs the end day event function and if it's day 7 then runs the code below
            if end_day_event(day, True) == True:
    
                statisticfunction("NEXT")
    
                if user_action(False, "You may continue playing as long as you want if you wish to do so!", 2, "CONTINUE", "EXIT") == 1:
                    day += 1

                else:
                    exit = True 

            else:
                day += 1

        # skip day
        elif main_menu_select == 3:

            if end_day_event(day, False) == True:
    
                statisticfunction("NEXT")

                if user_action(False, "You may continue playing as long as you want if you wish to do so!", 2, "CONTINUE", "EXIT") == 1:
                    day += 1
                    day_skipped += 1

                else:
                    exit = True

            else:
                day += 1
                day_skipped += 1

        # stats
        elif main_menu_select == 4:
            statisticfunction("BACK")

        # exit
        else:
            exit = True
```

```python
# RUN THIS BLOCK OF CODE AFTER THE TOP!!!!!! 

# opening screen + prologue
open_and_prologue()

# main menu
main_menu()

# user decided to exit
display(HTML("<h1>Thanks for playing!</h1>"))
```
